```python
from pathlib import Path
import copy
import json
import os
import platform
import random
import time

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import torch
from torch import nn
from torch.utils.data import DataLoader, Subset, TensorDataset
import torchvision
from torchvision.datasets import CIFAR10
from sklearn.model_selection import train_test_split

seed = 42
batch_size = 256
learning_rate = 0.001
weight_decay = 0.0001
max_epochs = 100
patience = 10
min_delta = 0.002
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
torch.set_num_threads(min(4, os.cpu_count() or 1))
os.environ['CUBLAS_WORKSPACE_CONFIG'] = ':4096:8'
torch.use_deterministic_algorithms(True)
torch.backends.cudnn.benchmark = False
output_dir = Path('hw7_results')
output_dir.mkdir(exist_ok=True)
plt.rcParams.update({'font.size': 10, 'figure.dpi': 120})


def set_seed():
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)


def synchronize():
    if device.type == 'cuda':
        torch.cuda.synchronize()


def make_loader(dataset, shuffle=False):
    generator = torch.Generator().manual_seed(seed)
    return DataLoader(dataset, batch_size=batch_size, shuffle=shuffle,
                      num_workers=0, generator=generator)


def build_network(hidden, activation):
    activation_class = {'ReLU': nn.ReLU, 'Sigmoid': nn.Sigmoid,
                        'Tanh': nn.Tanh}[activation]
    layers = [nn.Flatten()]
    inputs = 3 * 32 * 32
    for width in hidden:
        layers.extend([nn.Linear(inputs, width), activation_class()])
        inputs = width
    layers.append(nn.Linear(inputs, 10))
    return nn.Sequential(*layers).to(device)


def evaluate(model, loader):
    model.eval()
    total_loss, correct, count = 0.0, 0, 0
    loss_fn = nn.CrossEntropyLoss(reduction='sum')
    with torch.no_grad():
        for images, labels in loader:
            images, labels = images.to(device), labels.to(device)
            logits = model(images)
            total_loss += loss_fn(logits, labels).item()
            correct += (logits.argmax(1) == labels).sum().item()
            count += len(labels)
    return total_loss / count, correct / count


def save_plot(fig, name):
    fig.savefig(output_dir / name, dpi=180, bbox_inches='tight')
    plt.show()
    plt.close(fig)


def plot_history(key, title, history, selected_epoch):
    fig, axes = plt.subplots(1, 2, figsize=(10, 3.7),
                             constrained_layout=True)
    for ax, metric, ylabel in zip(
            axes, ['loss', 'accuracy'], ['Cross-entropy loss', 'Accuracy (%)']):
        factor = 100 if metric == 'accuracy' else 1
        ax.plot(history['epoch'], factor * history['training_' + metric],
                label='Training')
        ax.plot(history['epoch'], factor * history['validation_' + metric],
                label='Validation')
        ax.axvline(selected_epoch, color='gray', linestyle='--',
                   label='Selected epoch')
        ax.set(xlabel='Epoch', ylabel=ylabel)
        ax.grid(alpha=0.25)
        ax.legend(fontsize=8)
    fig.suptitle(title)
    save_plot(fig, key + '_curves.png')

def run_experiment(key, label, part, hidden, activation):
    set_seed()
    model = build_network(hidden, activation)
    optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate,
                                 weight_decay=weight_decay)
    loss_fn = nn.CrossEntropyLoss()
    train_loader = make_loader(fit_dataset, shuffle=True)
    history = []
    best_loss, stopping_reference = float('inf'), float('inf')
    best_state, selected_epoch, stale_epochs = None, 0, 0
    print('\n' + label)
    print('Hidden layers:', hidden, '| Activation:', activation)
    synchronize()
    started = time.perf_counter()
    for epoch in range(1, max_epochs + 1):
        model.train()
        for images, labels in train_loader:
            images, labels = images.to(device), labels.to(device)
            optimizer.zero_grad()
            loss = loss_fn(model(images), labels)
            loss.backward()
            optimizer.step()
        training_loss, training_accuracy = evaluate(model, fit_loader)
        validation_loss, validation_accuracy = evaluate(model, val_loader)
        history.append(dict(epoch=epoch, training_loss=training_loss,
                            training_accuracy=training_accuracy,
                            validation_loss=validation_loss,
                            validation_accuracy=validation_accuracy))
        if validation_loss < best_loss:
            best_loss, selected_epoch = validation_loss, epoch
            best_state = copy.deepcopy(model.state_dict())
        if validation_loss < stopping_reference - min_delta:
            stopping_reference, stale_epochs = validation_loss, 0
        else:
            stale_epochs += 1
        if epoch == 1 or epoch % 5 == 0 or stale_epochs >= patience:
            print(f'Epoch {epoch:3d}: train loss={training_loss:.4f}, '
                  f'train accuracy={training_accuracy:.2%}, '
                  f'validation loss={validation_loss:.4f}, '
                  f'validation accuracy={validation_accuracy:.2%}', flush=True)
        if stale_epochs >= patience:
            break
    synchronize()
    elapsed = time.perf_counter() - started
    model.load_state_dict(best_state)
    training_loss, training_accuracy = evaluate(model, fit_loader)
    validation_loss, validation_accuracy = evaluate(model, val_loader)
    test_loss, test_accuracy = evaluate(model, test_loader)
    row = dict(key=key, label=label, part=part, hidden=hidden,
               activation=activation,
               parameters=sum(p.numel() for p in model.parameters()
                              if p.requires_grad),
               epochs_trained=epoch, selected_epoch=selected_epoch,
               training_seconds=elapsed, seconds_per_epoch=elapsed / epoch,
               training_loss=training_loss, training_accuracy=training_accuracy,
               validation_loss=validation_loss,
               validation_accuracy=validation_accuracy,
               test_loss=test_loss, test_accuracy=test_accuracy,
               stop_reason='validation plateau' if stale_epochs >= patience
                           else '100-epoch limit')
    history = pd.DataFrame(history)
    history.to_csv(output_dir / (key + '_history.csv'), index=False)
    torch.save({k: v.cpu() for k, v in model.state_dict().items()},
               output_dir / (key + '_weights.pt'))
    predictions = []
    with torch.no_grad():
        for images, labels in test_loader:
            predictions.extend(model(images.to(device)).argmax(1).cpu().tolist())
    pd.DataFrame({'actual': test_labels.numpy(), 'predicted': predictions}).to_csv(
        output_dir / (key + '_test_predictions.csv'), index=False)
    results['experiments'].append(row)
    (output_dir / 'results.json').write_text(json.dumps(results, indent=2))
    print(f'Selected epoch: {selected_epoch}; epochs trained: {epoch}')
    print(f'Training time: {elapsed:.2f} seconds; '
          f'parameters: {row["parameters"]:,}')
    print(f'Selected model: train loss={training_loss:.4f}, '
          f'train accuracy={training_accuracy:.2%}, '
          f'test accuracy={test_accuracy:.2%}')
    plot_history(key, label, history, selected_epoch)
    return row


def compare_curves(keys, title, filename):
    fig, axes = plt.subplots(1, 2, figsize=(10, 3.7),
                             constrained_layout=True)
    for row in results['experiments']:
        if row['key'] not in keys:
            continue
        history = pd.read_csv(output_dir / (row['key'] + '_history.csv'))
        axes[0].plot(history['epoch'], history['validation_loss'],
                     label=row['label'])
        axes[1].plot(history['epoch'], 100 * history['validation_accuracy'],
                     label=row['label'])
    axes[0].set(ylabel='Validation cross-entropy loss')

    axes[1].set(ylabel='Validation accuracy (%)')
    for ax in axes:
        ax.set_xlabel('Epoch')
        ax.grid(alpha=0.25)
        ax.legend(fontsize=7)
    fig.suptitle(title)
    save_plot(fig, filename)


# %% CIFAR-10 data
set_seed()
train_data = CIFAR10(root='data', train=True, download=True)
test_data = CIFAR10(root='data', train=False, download=True)
train_labels = torch.tensor(train_data.targets, dtype=torch.long)
test_labels = torch.tensor(test_data.targets, dtype=torch.long)
fit_indices, val_indices = train_test_split(
    np.arange(len(train_labels)), test_size=0.10, random_state=seed,
    stratify=train_data.targets)
np.savez(output_dir / 'split_indices.npz', fit=fit_indices, validation=val_indices)
train_images = torch.from_numpy(train_data.data).permute(0, 3, 1, 2).float() / 255
test_images = torch.from_numpy(test_data.data).permute(0, 3, 1, 2).float() / 255
channel_sum = torch.zeros(3, dtype=torch.float64)
channel_square_sum = torch.zeros(3, dtype=torch.float64)
for indices in np.array_split(fit_indices, 90):
    batch = train_images[indices].double()
    channel_sum += batch.sum(dim=(0, 2, 3))
    channel_square_sum += (batch ** 2).sum(dim=(0, 2, 3))
pixel_count = len(fit_indices) * 32 * 32
channel_mean = (channel_sum / pixel_count).float()
channel_std = (channel_square_sum / pixel_count -
               (channel_sum / pixel_count) ** 2).sqrt().float()
train_images.sub_(channel_mean[None, :, None, None]).div_(
    channel_std[None, :, None, None])
test_images.sub_(channel_mean[None, :, None, None]).div_(
    channel_std[None, :, None, None])
train_images = train_images.contiguous()
test_images = test_images.contiguous()
full_train = TensorDataset(train_images, train_labels)
fit_dataset = Subset(full_train, fit_indices.tolist())
val_dataset = Subset(full_train, val_indices.tolist())
test_dataset = TensorDataset(test_images, test_labels)
fit_loader = make_loader(fit_dataset)
val_loader = make_loader(val_dataset)
test_loader = make_loader(test_dataset)
results = dict(
    metadata=dict(name='Anatoly Zelepouga', student_id='801236217', homework=7,
                  github='https://github.com/azelepou/IntroMachineLearning'),
    environment=dict(python=platform.python_version(), torch=torch.__version__,
                     torchvision=torchvision.__version__, numpy=np.__version__,
                     device=str(device), threads=torch.get_num_threads(),
                     processor=platform.processor() or platform.machine()),
    settings=dict(seed=seed, batch_size=batch_size, optimizer='Adam',
                  learning_rate=learning_rate, weight_decay=weight_decay,
                  max_epochs=max_epochs, patience=patience, min_delta=min_delta,
                  loss='CrossEntropyLoss',
                  timing='Wall time of training plus per-epoch training/validation '
                         'evaluation and checkpoint copies; excludes data preparation, '
                         'final testing, disk saves, and plotting.'),
    data=dict(official_training=50000, fitting=45000, validation=5000,
              official_test=10000, channel_mean=channel_mean.tolist(),
              channel_std=channel_std.tolist(), classes=train_data.classes,
              normalization='Channel mean/std computed on fitting samples only.',
              source='https://www.cs.toronto.edu/~kriz/cifar.html'),
    experiments=[])
print('HW7: Fully Connected Neural Networks for CIFAR-10')
print('Environment:', results['environment'])
print('Data: 45,000 fitting, 5,000 validation, 10,000 official test images.')
print('Channel mean:', channel_mean.tolist())
print('Channel standard deviation:', channel_std.tolist())
print('Each image is flattened to 3,072 inputs; outputs are 10 class logits.')

# %% Problem 1a
baseline = run_experiment('baseline', '1a: ReLU, 2 x 128', '1a',
                          [128, 128], 'ReLU')

# %% Problem 1b
four_layers = run_experiment('four_layers', '1b: ReLU, 4 x 128', '1b',
                             [128] * 4, 'ReLU')

# %% Problem 2a
ten_layers = run_experiment('ten_layers', '2a: ReLU, 10 x 128', '2a',
                            [128] * 10, 'ReLU')

# %% Problem 2b: Width
wide = run_experiment('wide', '2b: ReLU, 2 x 256', '2b', [256, 256], 'ReLU')

# %% Problem 2b: Sigmoid
sigmoid = run_experiment('sigmoid', '2b: Sigmoid, 2 x 128', '2b',
                         [128, 128], 'Sigmoid')

# %% Problem 2b: Tanh
tanh = run_experiment('tanh', '2b: Tanh, 2 x 128', '2b', [128, 128], 'Tanh')

# %% Comparisons
summary = pd.DataFrame(results['experiments'])
summary.to_csv(output_dir / 'summary.csv', index=False)
columns = ['label', 'parameters', 'epochs_trained', 'selected_epoch',
           'training_seconds', 'seconds_per_epoch', 'training_loss',
           'training_accuracy', 'validation_accuracy', 'test_accuracy']
print('\nAll experiments (accuracies shown as fractions):')
print(summary[columns].to_string(index=False, float_format=lambda x: f'{x:.4f}'))
compare_curves(['baseline', 'four_layers', 'ten_layers'],
               'Network depth comparison', 'depth_comparison.png')
compare_curves(['baseline', 'wide', 'sigmoid', 'tanh'],
               'Network width and activation comparison',
               'width_activation_comparison.png')
fig, axes = plt.subplots(1, 2, figsize=(10, 4), constrained_layout=True)
labels = ['2-layer\nReLU', '4-layer\nReLU', '10-layer\nReLU',
          'Wide\nReLU', 'Sigmoid', 'Tanh']
axes[0].bar(labels, summary['test_accuracy'] * 100)
axes[0].set(ylabel='Test accuracy (%)', title='Final evaluation')
axes[1].bar(labels, summary['training_seconds'])
axes[1].set(ylabel='Seconds', title='Training and monitoring time')
for ax in axes:
    ax.tick_params(axis='x', labelsize=8)
    ax.grid(axis='y', alpha=0.25)
save_plot(fig, 'summary_comparison.png')
print('\nSaved results, plots, predictions, and weights in', output_dir.resolve())
```
